```python
from sklearn.datasets import load_diabetes
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score
```

```python
X,y = load_diabetes(return_X_y=True)
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.2,random_state=2)
```

```python
Lreg = LinearRegression()
Lreg.fit(X_train,y_train)
y_predl = Lreg.predict(X_test)
print(r2_score(y_test, y_predl))
```

```python
Rreg = Ridge(alpha=0.1)
Rreg.fit(X_train,y_train)
y_predr = Rreg.predict(X_test)
print(r2_score(y_test, y_predr))
```

```python
Lareg = Lasso(alpha=0.01)
Lareg.fit(X_train, y_train)
y_predla = Lareg.predict(X_test)
print(r2_score(y_test, y_predla))
```

```python
Ereg = ElasticNet(alpha= 0.005, l1_ratio=0.9)
Ereg.fit(X_train, y_train)
y_prede = Ereg.predict(X_test)
print(r2_score(y_test, y_prede))
```
