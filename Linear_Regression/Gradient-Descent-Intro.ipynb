```python
import numpy as np
import pandas as pd

from sklearn.datasets import make_regression
```

```python
X,y = make_regression(n_samples=4, n_features=1, n_informative=1, n_targets=1, noise= 80, random_state=13)
```

```python
import matplotlib.pyplot as plt
plt.scatter(X,y)
```

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()

lr.fit(X,y)
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X), color='red')
```

```python
lr.coef_
```

```python
lr.intercept_
```

```python
m = 78.35
b = 100
```

```python
y_pred = ((78.35 * X) + b).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X), color='red', label='OLS')
plt.plot(X,y_pred,color='green', label='b=0')
plt.legend()
plt.show()
```

```python

loss_slope = -2 * np.sum(y - m*X.ravel() - b)

alfa = 0.1
step_size = loss_slope * alfa
b = b - step_size

y_pred1 = ((78.35 * X) + b).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X), color='red', label='OLS')
plt.plot(X,y_pred,color='lightgreen', label='b=0')
plt.plot(X,y_pred1,color='green', label='b={}'.format(b))
plt.legend()
plt.show()
```

```python
loss_slope = -2 * np.sum(y - m*X.ravel() - b)

alfa = 0.1
a = b

step_size = loss_slope * alfa
b = b - step_size

y_pred2 = ((78.35 * X) + b).reshape(4)
```

```python
plt.scatter(X,y)
plt.plot(X,lr.predict(X), color='red', label='OLS')
plt.plot(X,y_pred,color='lightgreen', label='b=0')
plt.plot(X,y_pred1,color='lightgreen', label=f'b={a}')
plt.plot(X,y_pred2,color='green', label='b={}'.format(b))
plt.legend()
plt.show()
```

```python
loss_slope = -2 * np.sum(y - m*X.ravel() - b)

alfa = 0.1
a1 = b
step_size = loss_slope * alfa
b = b - step_size

y_pred3 = ((78.35 * X) + b).reshape(4)

plt.scatter(X,y)
plt.plot(X,lr.predict(X), color='red', label='OLS')
plt.plot(X,y_pred,color='lightgreen', label='b=0')
plt.plot(X,y_pred1,color='lightgreen', label=f'b={a}')
plt.plot(X,y_pred2,color='lightgreen', label=f'b={a1}')
plt.plot(X,y_pred3,color='green', label='b={}'.format(b))
plt.legend()
plt.show()
```

```python
b = -100
m = 78.35
alfa = 0.1
epochs = 10

for i in range(epochs):
    loss_slope = -2 * np.sum(y - m*X.ravel() - b)
    b = b - (alfa*loss_slope)

    y_pred  = m * X + b

    plt.plot(X, y_pred)

plt.scatter(X , y)
```
