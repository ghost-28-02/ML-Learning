```python
from sklearn.datasets import make_classification
import numpy as np 
import pandas as pd
X,y = make_classification(n_samples=100, n_features=2, n_informative=1, n_redundant=0, n_classes=2, n_clusters_per_class=1, hypercube=False, class_sep=10, random_state=41)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(13,6))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def step(z):
    return 1 if z > 0 else 0
```

```python
def perceptron(X,y):
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.1

    for i in range(100):
        j = np.random.randint(0,100)
        y_hat = step(np.dot(X[j],weights))
        error = y[j] - y_hat
        weights += lr*error*X[j]

    return weights[0], weights[1:]
```

```python
intercept_, coef_ = perceptron(X,y)
```

```python
print(coef_)
print(intercept_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
plt.figure(figsize=(13,5))
plt.plot(x_input, y_input,color='r', linewidth = 3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression()
lr.fit(X,y)
```

```python
m = -(lr.coef_[0][0]/lr.coef_[0][1])
b = -(lr.intercept_/lr.coef_[0][1])
```

```python
x_input1 = np.linspace(-3,3,100)
y_input1 = m*x_input1 + b
```

```python
plt.figure(figsize=(13,5))
plt.plot(x_input, y_input,color='r', linewidth = 3)
plt.plot(x_input1, y_input1,color='black', linewidth = 3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
