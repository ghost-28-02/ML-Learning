```python
import numpy as np
```

```python
import pickle

with open('models/ohe_sex.pkl', 'rb') as f:
    ohe_sex = pickle.load(f)

with open('models/ohe_embarked.pkl', 'rb') as f:
    ohe_embarked = pickle.load(f)

with open('models/clf.pkl', 'rb') as f:
    clf = pickle.load(f)
```

```python
test_input = np.array([2,'male',31.0,0,0,10.5,'S'], dtype=object).reshape(1,7)
```

```python
test_input_sex = ohe_sex.transform(test_input[:,1].reshape(1,1))
```

```python
test_input_embarked = ohe_embarked.transform(test_input[:,-1].reshape(1,1))
```

```python
test_input_age = test_input[:,2].reshape(1,1)
```

```python
test_input_transform = np.concatenate((test_input[:,[0,3,4,5]],test_input_sex,test_input_age,test_input_embarked),axis=1)
```

```python
clf.predict(test_input_transform)
```
