```python
for j in range(1, 100):
    url = f'https://www.ambitionbox.com/list-of-companies?campaign=desktop_nav&page={j}'
    print(f"Fetching page {j}")
    webpag = session.get(url, headers=headers).text
    time.sleep(2)

    soup = BeautifulSoup(webpag, 'html.parser')
    companies = soup.find_all('div', class_='companyCardWrapper')

    name = []
    rating = []
    voting = []
    ctype = []

    for company in companies:
        try:
            name.append(company.find('h2').text.strip())
        except:
            name.append(None)
        try:
            rating.append(company.find('div', class_='rating_text').text.strip())
        except:
            rating.append(None)
        try:
            voting.append(company.find('span', class_='companyCardWrapper__companyRatingCount').text.strip())
        except:
            voting.append(None)
        try:
            ctype.append(company.find('span', class_='companyCardWrapper__interLinking').text.strip())
        except:
            ctype.append(None)

    df = pd.DataFrame({
        'name': name,
        'rating': rating,
        'voting': voting,
        'type': ctype
    })

    final = pd.concat([final, df], ignore_index=True)
```
