```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
X = np.array([[1,2],[2,2],[2,3],[8,7],[8,8],[25,80]])
```

```python
plt.scatter(X[:,0],X[:,1])
```

```python
from sklearn.cluster import DBSCAN
db = DBSCAN(min_samples=2, eps=3)
db.fit(X)
```

```python
db.labels_
```

```python
from sklearn.datasets import make_circles

X,y = make_circles(n_samples=500, factor=0.5,noise=0.03,random_state=28)

dbscan = DBSCAN(min_samples=5,eps=0.3)
clusters = dbscan.fit_predict(X)

plt.scatter(X[:,0], X[:,1], c=clusters, s=10, cmap='winter')
```
