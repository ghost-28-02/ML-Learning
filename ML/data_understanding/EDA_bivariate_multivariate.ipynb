### 2. Bar Plot(Numerical - Categorical)

```python
titanic.head()
```

```python
sns.barplot(x='Pclass',y='Age',palette=['blue','orange','green'],hue='Sex',data=titanic)
```

### 3. Box Plot (Numerical - Categorical)

```python
sns.boxplot(x='Sex',y='Age',hue='Survived',data=titanic)
```

### 4. Displot (Numerical - Categorical)

```python
sns.kdeplot(data=titanic[titanic['Survived'] == 0]['Age'], label='Did not survive')
sns.kdeplot(data=titanic[titanic['Survived'] == 1]['Age'], label='Survived')

plt.xlabel('Age')
plt.ylabel('Density')
plt.title('Age Distribution by Survival Status')
plt.legend()
plt.show()

```

### 5. HeatMap (Categorical - Categorical)

```python
sns.heatmap(pd.crosstab(titanic['Pclass'],titanic['Survived']))
```

```python
(titanic.groupby('Sex')[['Survived']].mean()*100)
```

### 6. ClusterMap (Categorical - Categorical)

```python
sns.clustermap(pd.crosstab(titanic['Parch'],titanic['Survived']))
```
