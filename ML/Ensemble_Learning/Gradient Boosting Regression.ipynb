```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.DataFrame([
    [165,137,472,192],
    [101,92,250,144],
    [29,127,201,91]
],
columns=['R&D Spend','Administration','Markiting Spend','Profit'])
```

```python
df.head()
```

```python
df['f0(x)'] = df['Profit'].mean()
df
```

```python
df['r1'] = df['Profit'] - df['f0(x)']
df
```

```python
from sklearn.tree import DecisionTreeRegressor
dt1 = DecisionTreeRegressor(max_depth=1)
```

```python
dt1.fit(df.iloc[:,:3].values, df['r1'].values)
```

```python
from sklearn.tree import plot_tree
plot_tree(dt1)
```
