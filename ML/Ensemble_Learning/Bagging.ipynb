```python
import numpy as np
import pandas as pd
```

```python
df = pd.read_csv('/Users/sachinchoudhary/Developer/ML/data/iris.csv')
df.head()
```

```python
from sklearn.preprocessing import LabelEncoder
encoder = LabelEncoder()

df['class'] = encoder.fit_transform(df['class'])
```

```python
df = df[df['class'] != 0][['sepal_width', 'petal_length', 'class']]
```

```python
df.head()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.scatter(df['sepal_width'],df['petal_length'], c=df['class'], cmap='winter')
```

```python
df = df.sample(100)
df_train = df.iloc[:60,:].sample(10)
df_val = df.iloc[60:80,:].sample(5)
df_test = df.iloc[80:,:].sample(5)
```

```python
X_test = df_val.iloc[:,:2].values
y_test = df_val.iloc[:,-1].values
```

# Case 1 - Bagging

```python
df_bag = df_train.sample(8, replace=True)
X = df_bag.iloc[:,:2]
y = df_bag.iloc[:,-1]

df_bag
```

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
from mlxtend.plotting import plot_decision_regions
from sklearn.metrics import accuracy_score
```

```python
def evaluate(clf,X,y):
    clf.fit(X,y)
    plot_tree(clf)
    plt.show()
    plot_decision_regions(X.values, y.values,clf=clf,legend=2)
    y_pred = clf.predict(X_test)
    print('accuracy ->',accuracy_score(y_test, y_pred))
```

```python
dt_bag1 = DecisionTreeClassifier()
evaluate(dt_bag1,X,y)
```

```python
df_bag = df_train.sample(8, replace=True)
X = df_bag.iloc[:,:2]
y = df_bag.iloc[:,-1]

df_bag
```

```python
dt_bag2 = DecisionTreeClassifier()
evaluate(dt_bag2,X,y)
```

```python
df_bag = df_train.sample(8, replace=True)
X = df_bag.iloc[:,:2]
y = df_bag.iloc[:,-1]

df_bag
```

```python
dt_bag3 = DecisionTreeClassifier()
evaluate(dt_bag3,X,y)
```
