```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/Users/sachinchoudhary/Developer/ML/data/student_clustering.csv')
df.head()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(df['cgpa'], df['iq'])
```

```python
from sklearn.cluster import KMeans

wcss = []

for i in range(1,21):
    km = KMeans(n_clusters=i)
    km.fit(df)
    wcss.append(km.inertia_)
```

```python
wcss
```

```python
plt.plot(range(1,21), wcss)
```

```python
X = df.iloc[:,:].values
km = KMeans(n_clusters=4)
y_mean = km.fit_predict(X)
```

```python
# Plot the clustered data points with their cluster assignments
plt.figure(figsize=(8,6))
plt.scatter(X[:, 0], X[:, 1], c=y_mean, cmap='viridis', edgecolor='k', s=60)
plt.xlabel('CGPA')
plt.ylabel('IQ')
plt.title('K-Means Clustering (k=4)')
plt.colorbar(label='Cluster')
plt.show()
```

# 3D

```python
from sklearn.datasets import make_blobs
```

```python
centroid = [(-5,-5,5),(5,5,-5),(3.5,-2.5,4),(-2.5,2.5,-4)]
cluster_std = [1,1,1,1]

X,y = make_blobs(n_samples=200,cluster_std=cluster_std,centers=centroid,n_features=3,random_state=28)
```

```python
import plotly.express as px
fig = px.scatter_3d(x=X[:,0],y=X[:,1],z=X[:,2])
fig.show()
```

```python
wcss = []

for i in range(1,21):
    km = KMeans(n_clusters=i)
    km.fit(X)
    wcss.append(km.inertia_)

plt.plot(range(1,21), wcss)
```

```python
km = KMeans(n_clusters=4)
y_pred = km.fit_predict(X)
```

```python
df = pd.DataFrame()

df['col1'] = X[:,0]
df['col2'] = X[:,1]
df['col3'] = X[:,2]
df['label'] = y_pred
```

```python
fig = px.scatter_3d(df,x='col1',y='col2',z='col3',color='label')
fig.show()
```
