```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv('/Users/sachinchoudhary/Developer/ML/data/heart.csv')
df.head()
```

```python
X = df.iloc[:,:-1]
y = df.iloc[:,-1]
```

```python
X_train, X_test, y_trian, y_test = train_test_split(X,y,test_size=0.2,random_state=28)
```

```python
rf = RandomForestClassifier()
rf.fit(X_train, y_trian)
y_pred = rf.predict(X_test)
print(accuracy_score(y_test, y_pred))
```

```python
rf = RandomForestClassifier(max_samples=0.75, random_state=28)
rf.fit(X_train, y_trian)
y_pred = rf.predict(X_test)
print(accuracy_score(y_test, y_pred))
```

```python
from sklearn.model_selection import cross_val_score
print(np.mean(cross_val_score(RandomForestClassifier(),X,y,scoring='accuracy',cv=10)))
```

# GridSearchCv

```python
param = {
    'n_estimators': [20,60,100,120],
    'max_features': [0.2,0.6,1.0],
    'max_depth' : [2,8,None],
    'max_samples' : [0.25,0.5,0.75,1.0]
}
```

```python
rf = RandomForestClassifier()
```

```python
from sklearn.model_selection import GridSearchCV
grid = GridSearchCV(estimator=rf,param_grid=param, cv=10, n_jobs=-1, verbose=2)
grid.fit(X_train,y_trian)
```

```python
grid.best_params_
```

```python
grid.best_score_
```

```python
param = {
    'n_estimators': [20,60,100,120],
    'max_features': [0.2,0.6,1.0],
    'max_depth' : [2,8,None],
    'max_samples' : [0.25,0.5,0.75,1.0],
    'bootstrap': [True, False],
    'min_samples_split': [2,5],
    'min_samples_leaf':[1,2]
}
```

```python
from sklearn.model_selection import RandomizedSearchCV
rf = RandomForestClassifier()
```

```python
rsv = RandomizedSearchCV(rf,param_distributions=param,cv=10,n_jobs=-1,verbose=2,)
```

```python
rsv.fit(X_train, y_trian)
```

```python
rsv.best_params_
```
