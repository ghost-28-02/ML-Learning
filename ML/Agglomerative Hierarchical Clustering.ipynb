```python
import pandas as pd 
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv('/Users/sachinchoudhary/Developer/ML/data/hierarchical-clustering-with-python-and-scikit-learn-shopping-data.csv')
df.head()
```

```python
data = df.iloc[:,3:5].values
data
```

```python
import scipy.cluster.hierarchy as shc
plt.figure(figsize=(13,6))
plt.title('Customer Dendograms')
dend = shc.dendrogram(shc.linkage(data,method='ward'))
```

```python
from sklearn.cluster import AgglomerativeClustering
clusters = AgglomerativeClustering(n_clusters=5, linkage='ward')
labels_ = clusters.fit_predict(data)
```

```python
plt.figure(figsize=(13,6))
plt.scatter(data[:,0],data[:,1],c=clusters.labels_,cmap='rainbow')
```
