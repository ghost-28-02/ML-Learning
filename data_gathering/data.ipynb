```python
import numpy as np
import sklearn.datasets as sd
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split 
```

```python
df = pd.DataFrame(dataset.data,columns=dataset.feature_names)
df.head()
```

```python
df.shape
```

```python
x = df
y = dataset.target
```

```python
print(y)
```
