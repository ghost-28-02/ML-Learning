```python
for i in range(1,3):
    response = requests.get('https://api.themoviedb.org/3/movie/top_rated?api_key=271041db3459fcfc7b4816c92254c9ce&language=en-US&page={}'.format(i))
    temp_df = pd.DataFrame(response.json()['results'])[['id','title','overview','release_date','popularity','vote_average','vote_count']]
    df = pd.concat([df, temp_df], ignore_index=True)
```

```python
for i in range(1, 508):
    try:
        response = requests.get(f'https://api.themoviedb.org/3/movie/top_rated?api_key=271041db3459fcfc7b4816c92254c9ce&language=en-US&page={i}')
        if response.status_code == 200:
            data = response.json()['results']
            temp_df = pd.DataFrame(data)[['id', 'title', 'overview', 'release_date', 'popularity', 'vote_average', 'vote_count']]
            df = pd.concat([df, temp_df], ignore_index=True)
        else:
            print(f"Failed to fetch page {i}: Status code {response.status_code}")
        
        time.sleep(0.3)  # sleep for 300ms to stay under rate limit
    except requests.exceptions.RequestException as e:
        print(f"Request failed at page {i}: {e}")
        time.sleep(2)
```
