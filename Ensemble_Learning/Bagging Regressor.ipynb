```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/Users/sachinchoudhary/Developer/ML/data/BostonHousing.csv')
```

```python

df['rm'].fillna(df['rm'].mean(),inplace=True)
```

```python
X = df.iloc[:,:13]
y = df.iloc[:,-1]
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.neighbors import KNeighborsRegressor
from sklearn.model_selection import GridSearchCV
from sklearn.tree import DecisionTreeRegressor
from sklearn.metrics import r2_score
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,y, test_size=0.2, random_state=28)
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
knn = KNeighborsRegressor()
```

```python
lr.fit(X_train, y_train)
dt.fit(X_train, y_train)
knn.fit(X_train, y_train)
```

```python
y_pred1 = lr.predict(X_test)
y_pred2 = dt.predict(X_test)
y_pred3 = knn.predict(X_test)
```

```python
print("r2_score for LR ->",r2_score(y_test, y_pred1))
print("r2_score for DT ->",r2_score(y_test, y_pred2))
print("r2_score for KNN ->",r2_score(y_test, y_pred3))
```

```python
from sklearn.ensemble import BaggingRegressor
```

```python
bag = BaggingRegressor(random_state=28)
bag.fit(X_train, y_train)
y_pred = bag.predict(X_test)

print(bag.score(X_train, y_train))
print(bag.score(X_test, y_test))
```
