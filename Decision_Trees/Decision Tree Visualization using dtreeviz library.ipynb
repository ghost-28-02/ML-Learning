# 10. Univariate Regression

```python
from sklearn.tree import DecisionTreeRegressor
from dtreeviz import rtreeviz_univar

df_cars = pd.read_csv("/Users/sachinchoudhary/Developer/ML/data/auto-mpg.csv", usecols=['mpg','cylinders','weight','displacement'])
X, y = df_cars[['weight']], df_cars['mpg']

dt = DecisionTreeRegressor(max_depth=3, criterion="absolute_error")
dt.fit(X, y)

fig = plt.figure()
ax = fig.gca()
rtreeviz_univar(dt, X, y, 'weight', 'mpg', ax=ax)
plt.show()
```

# 11. 3-D Regression

```python
from sklearn.tree import DecisionTreeRegressor
from dtreeviz import rtreeviz_bivar_3D
import matplotlib.pyplot as plt

# Prepare data
X = df_cars[['weight', 'displacement']]
y = df_cars['mpg']

# Train decision tree regressor
dt = DecisionTreeRegressor(max_depth=3, criterion="absolute_error")
dt.fit(X, y)

# Create 3D plot
fig = plt.figure(figsize=(6, 5))
ax = fig.add_subplot(111, projection='3d')

# Visualize the regression tree
rtreeviz_bivar_3D(
    tree_model=dt,
    X_train=X,
    y_train=y,
    feature_names=['Vehicle Weight', 'Displacement'],
    target_name='MPG',
    ax=ax,
    fontsize=14,
    azim=25,
    elev=20,
    dist=8.2,
    show=['splits', 'title']
)

plt.show()
```
