```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(z):
    return 1/(1 + np.exp(-z))
```

```python
from sklearn.datasets import make_classification
X,y = make_classification(n_samples=100, n_features=2, n_informative=1,n_redundant=0, n_classes=2, n_clusters_per_class=1,random_state=41,class_sep=10,hypercube=False)
```

```python
plt.figure(figsize=(13,4))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression(penalty=None, solver='sag')
lr.fit(X,y)
```

```python
print(lr.coef_)
print(lr.intercept_)
```

```python
m = -(lr.coef_[0][0]/lr.coef_[0][1])
b = -(lr.intercept_/lr.coef_[0][1])
```

```python
x_input = np.linspace(-3,3,100)
y_input = m*x_input + b
```

```python
plt.figure(figsize=(13,5))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.plot(x_input,y_input,color='r')
plt.ylim(-3,2)
```

```python
def sigmoid(z):
    return 1/(1 + np.exp(-z))
```

```python
def gd(X,y):
    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    lr = 0.3

    for i in range(10000):
        y_hat = sigmoid(np.dot(X,weights))
        weights += lr*np.dot((y-y_hat),X)/X.shape[0]

    return weights[0], weights[1:]
```

```python
intercept_, coef_ = gd(X,y)
```

```python
print(intercept_)
print(coef_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])
```

```python
x_input1 = np.linspace(-3,3,100)
y_input1 = m*x_input1 + b
```
