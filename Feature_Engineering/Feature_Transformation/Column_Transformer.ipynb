```python
import pandas as pd
import numpy as np
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OrdinalEncoder
from sklearn.preprocessing import OneHotEncoder
```

```python
df = pd.read_csv('/Users/sachinchoudhary/Developer/ML/data/covid_toy.csv')
```

```python
df.sample(5)
```

```python
from sklearn.model_selection import train_test_split    
X_train, X_test, y_train, y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'], test_size=0.2)
```

# 1. Aam Zindagi (😢)

```python
si = SimpleImputer()

X_train_fever = si.fit_transform(X_train[['fever']])
X_test_fever = si.fit_transform(X_test[['fever']])
```

```python
oe = OrdinalEncoder(categories=[['Mild', 'Strong']])

X_train_cough = oe.fit_transform(X_train[['cough']])
X_test_cough = oe.fit_transform(X_test[['cough']])
```

```python
oht = OneHotEncoder(drop='first', sparse_output=False)
X_train_gender_city = oht.fit_transform(X_train[['gender', 'city']])
X_test_gender_city = oht.fit_transform(X_test[['gender', 'city']])
```

```python
X_train_age = X_train.drop(columns=['gender', 'fever', 'cough', 'city']).values
X_test_age = X_test.drop(columns=['gender', 'fever', 'cough', 'city']).values
```

```python
X_train_transform = np.concatenate((X_train_age,X_train_fever,X_train_gender_city,X_train_cough), axis=1)
X_test_transform = np.concatenate((X_test_age,X_test_fever,X_test_gender_city,X_test_cough), axis=1)
```

# 2. Mentos Zindagi (😎)

```python
from sklearn.compose  import ColumnTransformer
transformer = ColumnTransformer(transformers=[
    ('tnf1',SimpleImputer(),['fever']),
    ('tnf2',OrdinalEncoder(categories=[['Mild', 'Strong']]),['cough']),
    ('tnf3',OneHotEncoder(drop='first', sparse_output=False),['gender', 'city'])
], remainder='passthrough')
```

```python
X_train_transform = transformer.fit_transform(X_train)
X_test_transform = transformer.transform(X_test)
```
