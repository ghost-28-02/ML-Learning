```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
from sklearn.model_selection import cross_val_score
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv('/Users/sachinchoudhary/Developer/ML/data/train.csv', usecols=['Age','Pclass','SibSp','Survived','Parch'])
df.head()
```

```python
df.dropna(inplace=True)
```

```python
X = df.iloc[:,1:]
y = df['Survived']
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
```

# Applying Feature Construction

```python
X['Family_size'] = X['SibSp'] + X['Parch'] + 1
X.sample(5)
```

```python
def myfxn(num):
    if num <= 1:
        return 0
    elif num > 1 & num <= 4:
        return 1
    else:
        return 2
    
```

```python
X['Family_type'] = X['Family_size'].apply(myfxn)
```

```python
X.head()
```

```python
X.drop(columns=['SibSp','Parch','Family_size'], inplace=True)
X.sample(5)
```

```python
np.mean(cross_val_score(LogisticRegression(),X,y,scoring='accuracy',cv=20))
```
